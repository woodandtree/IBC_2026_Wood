```python
import pickle
# load dictionary with genetic code from pickle file
genetic_code = pickle.load(open("genetic_code.pickle", "rb")) #beginning stays the same

def get_amino_acids(mRNA): #use seq instead of mRNA to make it the same as how we learned in class
    mRNA = mRNA.replace("T", "U") #this replaces all the Ts in the code to Us
    
    i = 0
    aa_sequence = []
    while (i + 3) < len(mRNA):
        codon = mRNA[i:(i + 3)]
        aa = genetic_code[codon]
            
        if aa == "Stop":
            break
        else:
            aa_sequence.append(aa)
            i = i + 3 
        
    return "".join(aa_sequence)

    #same code from the previous question

with open("Turkey_transcripts_15_coding.fasta","r") as infile, open("Turkey_15.fasta","w") as outfile:
#add the input file and tell it to write the output file

    for line in infile: #for each line in the infile
        line = line.strip() #strips the white space
        
        if line.startswith(">"): #all the seperate fasta sequences start with >
            end = line.replace("gbkey=CDS", "gbskey=AA") #replace the gbskey=CDS with gbskey=AA, need to give it a name so I can tell it to write it into the outfile
            #^this was not working because its gbkey=CDS not gbskey=CDS needs to be exact
            outfile.write(end + "\n") #this says write the new header in and then move to the next line(cause otherwise it looks all jumbled together) BUT if next line not start with > then need an else arguement

        else: 
            seq = get_amino_acids(line) #name it seq then for each line that's not > it will run get_amino_acids for that line 
            outfile.write(seq + "\n") #write the seq in and then go to the next line
   
```
