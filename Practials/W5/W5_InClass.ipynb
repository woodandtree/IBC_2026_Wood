```python
with open("some_seqs.txt", "r") as infile, open("gc_content.txt","w") as outfile:
    for line in infile:
        line = line.rstrip()
        gene_id,seq = line.split()
        gc = seq.count("G") + seq.count("C") / len(seq)
        outfile.write(gene_id+"\t"+str(gc)+"\n")
```
