```python
with open("Turkey_transcripts_15.fasta","r") as infile, open("gc_contentturkey.txt","w") as outfile:
    seq = ""
    name = ""
    for line in infile: 
        line = line.rstrip()
        # can't use because not seprated by tabs code, seq = line.split() 
        # but all new line of the code starts with > maybe I can do something with that
        if line.startswith(">"):
            if seq:
                name = line.split()
                gc = (seq.count("G") + seq.count("C")) / len(seq)
                outfile.write(name+"\t"+str(gc)+"\n")
```
